```python
# Числовые признаки
num_cols = [
    'tenure',
    'MonthlyCharges',
    'TotalCharges'
]

# Категориальные признаки
cat_cols = [
    'gender',
    'Partner',
    'Dependents',
    'PhoneService',
    'MultipleLines',
    'InternetService',
    'OnlineSecurity',
    'OnlineBackup',
    'DeviceProtection',
    'TechSupport',
    'StreamingTV',
    'StreamingMovies',
    'Contract',
    'PaperlessBilling',
    'PaymentMethod'
]
```

```python
X = df.drop(columns=['Churn'])
y = df['Churn'].values

from sklearn.model_selection import train_test_split

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
    stratify=y
)

print(type(X_train))
print(type(X_test))
```

#### One-Hot Encoding и StandardScaler

```python
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import StandardScaler, OneHotEncoder

# 'num' — произвольное название нашего преобразования
# StandardScaler() — объект, который будет стандартизировать данные
# num_cols — список столбцов, к которым нужно применить стандартизацию
# handle_unknown='ignore'- если встретилась неизвестная категория, не выдавай ошибку, а закодируй её нулевым вектором для соответствующего признака
# sparse_output=False - означает, что нужно вернуть обычный плотный массив NumPy, а не разреженную матрицу
# remainder='passthrough' - все оставшиеся столбцы передать дальше без изменений
# remainder='drop'- то все неуказанные столбцы были бы удалены

preprocessor = ColumnTransformer(
    transformers=[
        ('num', StandardScaler(), num_cols),
        ('cat', OneHotEncoder(
            handle_unknown='ignore',
            sparse_output=False
        ), cat_cols)
    ],
    remainder='passthrough'
)


X_train = preprocessor.fit_transform(X_train)
X_test = preprocessor.transform(X_test)
```

```python
print('Размер X_train:', X_train.shape)
print('Размер X_test:', X_test.shape)

print('Тип X_train:', X_train.dtype)

print('NaN в X_train:', np.isnan(X_train).sum())
print('NaN в X_test:', np.isnan(X_test).sum())

print('Бесконечные значения:', np.isinf(X_train).sum())
```

## Модель

```python
class LogisticRG:
    """  Параметры
    ------------
    eta : float
        Скорость обучения (между 0.0 и 1.0)
    n_iter : int
        Кол-во проходов по обуучающей выборке
    random_state : int
        Опорные значения генератора случайных чисел для инициализации весов.

    Атрибуты
    ------------
    w_ : 1d-array 
        Одномерный массив. Веса после подгонки
    b_ : Scalar
        Смещение после подгонки
    errors_ : list
        Кол-во неправильных классификация(обюновлений) в каждой эпохе.
    """

    def __init__(self, eta=0.01, n_iter=50, random_state=1):
        self.eta = eta
        self.n_iter = n_iter
        self.random_state = random_state

    def fit(self, X, y):
        """ Соответствие тренировочным данным.

        Параметры
        ------------
        X : {array-like}, shape = [n_examples, n_features]
            Обучающий вектор, где:
                                    n_examples - кол-во образцов,
                                   | n_features - кол-во признаков
        y : array-like, shape = [n_examples]
            Целевые значения
            
        Возвращаем значения
        ------------
        self : object
        """
        rgen = np.random.RandomState(self.random_state)  # генерация случайных чисел для весов
        self.w_ = rgen.normal(loc=0.0, scale=0.01, size=X.shape[1])  # Подсчет начальных весов через Гассовское распределения (нормальное распределение)


        self.b_ = 0.0  # Инициализируем начальное смещение
        self.losses_ = []  # Инициализируем пустой список для хранения ошибок
        eps = 1e-15

        # Основой цикл обучения по эпохам
        for i in range(self.n_iter):
    
            # Линейная комбинация
            net_input = self.net_input(X)
    
            # Вероятности
            output = self.activation(net_input)
    
            # Ошибки
            errors = y - output
    
            # Обновление весов
            self.w_ += self.eta * X.T.dot(errors) / X.shape[0]
            self.b_ += self.eta * errors.mean()
    
            # Пересчитываем вероятности после обновления весов
            output = self.predict_proba(X)
            output = np.clip(output, eps, 1 - eps)
    
            # Binary Cross-Entropy
            loss = (
                -y.dot(np.log(output))
                - (1-y).dot(np.log(1-output))
            ) / X.shape[0]
    
            self.losses_.append(loss)
    
        return self

    def net_input(self, X):
        """ Линейная функция расчета значения входа """
        return np.dot(X, self.w_) + self.b_  # Расчет взвешенной суммы z=w_1x_1+w_2x_2 ... +b
        
    def activation(self, z): 
        """ Вычисление логистической сигмоидной активации """
        return 1.0 / (1.0 + np.exp(-np.clip(z, -250, 250)))
        
    def predict_proba(self, X):
        """Возвращает вероятность принадлежности к классу 1."""
        return self.activation(self.net_input(X))
        
    def predict(self, X):
        """ Функция предсказания """
        probabilities = self.predict_proba(X)
        return np.where(probabilities >= 0.4, 1, 0)
```

## Обучение модели и оценка качества

```python
model = LogisticRG(
    eta=0.01,
    n_iter=1000,
    random_state=1
)
model.fit(X_train, y_train)
```

```python
y_proba = model.predict_proba(X_test)
y_pred = model.predict(X_test)
```

```python
print(accuracy_score(y_test, y_pred))
print(classification_report(y_test, y_pred))
```

```python
plt.figure(figsize=(10, 5))

plt.plot(model.losses_)

plt.xlabel('Epoch')
plt.ylabel('Binary Cross-Entropy')
plt.title('Training Loss')
plt.grid()
plt.show()

print('Начальная ошибка:', model.losses_[0])
print('Конечная ошибка:', model.losses_[-1])
```

```python
from sklearn.metrics import classification_report
import numpy as np

y_proba = model.predict_proba(X_test)

for threshold in [0.3, 0.4, 0.5, 0.6, 0.7]:
    y_pred_threshold = (y_proba >= threshold).astype(int)

    print(f'\nПорог: {threshold}')
    print(classification_report(
        y_test,
        y_pred_threshold,
        target_names=['No', 'Yes'],
        zero_division=0
    ))
```

```python
from sklearn.metrics import precision_score, recall_score, f1_score

thresholds = np.arange(0.1, 0.91, 0.01)

precisions = []
recalls = []
f1_scores = []

for threshold in thresholds:

    y_pred = np.where(y_proba >= threshold, 1, 0)

    precisions.append(precision_score(y_test, y_pred, zero_division=0))
    recalls.append(recall_score(y_test, y_pred, zero_division=0))
    f1_scores.append(f1_score(y_test, y_pred, zero_division=0))

plt.figure(figsize=(12, 6))

plt.plot(thresholds, precisions, label='Precision')
plt.plot(thresholds, recalls, label='Recall')
plt.plot(thresholds, f1_scores, label='F1-score')

plt.axvline(0.4, linestyle='--', label='Threshold = 0.4')

plt.xlabel('Пороги')
plt.ylabel('Показатели качества')
plt.title('Precision, Recall и F1-score vs Threshold')

plt.legend()
plt.grid()
plt.show()
```
